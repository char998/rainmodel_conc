```python
import numpy as np

def kessler_precipitation(F_c, dt_s, q_l0=1.0, k=1e-3):
    """
    Compute precipitation from condensation flux using Kessler autoconversion.

    Parameters
    ----------
    F_c : array-like
        Condensation flux (kg/m2/s) at each timestep.
    dt_s : float
        Timestep in seconds.
    q_l0 : float
        Threshold LWP (kg/m2) for autoconversion.
    k : float
        Autoconversion rate (s^-1).

    Returns
    -------
    P_precip : array
        Precipitation flux (kg/m2/s) at each timestep.
    LWP : array
        Cloud liquid water path accumulated over timestep (kg/m2).
    """
    F_c = np.array(F_c)
    # Accumulated LWP in this timestep (kg/m2)
    LWP = F_c * dt_s
    
    # Apply Kessler: only water above threshold converts to rain
    excess_LWP = np.maximum(0.0, LWP - q_l0)
    # Precipitation produced in this timestep
    P_precip = k * excess_LWP  # kg/m2/s, same units as flux

    # Convert to mm/hr for easy comparison
    P_mm_hr = P_precip * 3600.0

    return P_precip, P_mm_hr, LWP

# Example usage
F_c = np.array([0.0, 0.0, 0.0, 0.00131455, 0.01212818, 0.01709459,
                0.00753242, 0.0, 0.0024968, 0.00569086, 0.00879037,
                0.00097028, 0.0, 0.0])

dt = 3600.0  # 1 hour timestep

P_flux, P_mm_hr, LWP = kessler_precipitation(F_c, dt, q_l0=1.0, k=1e-3)

for i, p in enumerate(P_mm_hr):
    print(f"Step {i}: LWP = {LWP[i]:.2f} kg/m2, Precip = {p:.2f} mm/hr")
```
